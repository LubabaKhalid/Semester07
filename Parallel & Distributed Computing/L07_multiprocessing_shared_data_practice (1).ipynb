## Section 1 – Shared List (Practice)

```python

from multiprocessing import Process, Manager

def worker_append(shared_list, value):
    shared_list.append(value)

if __name__ == "__main__":
    with Manager() as manager:
        shared_list = manager.list()
        processes = []
        for i in range(5):
            p = Process(target=worker_append, args=(shared_list, i))
            processes.append(p)
            p.start()
        for p in processes:
            p.join()
        print("Final shared list:", list(shared_list))
```

```python
from multiprocessing import Process, Manager

def worker_append(shared_list, value):
    shared_list.append(value)

if __name__ == "__main__":
    with Manager() as manager:
        shared_list = manager.list()

        processes = [Process(target=worker_append, args=(shared_list, i)) for i in range(10)]

        for p in processes: p.start()
        for p in processes: p.join()

        print("Final shared list:", list(shared_list))
```

## Section 3 – Shared Complex Object (Practice)

```python

from multiprocessing import Process, Manager

def worker_scores(shared_data, student, score):
    if student not in shared_data:
        shared_data[student] = []
    shared_data[student].append(score)

if __name__ == "__main__":
    with Manager() as manager:
        shared_data = manager.dict()
        processes = []
        tasks = [("Alice", 85), ("Bob", 90), ("Alice", 95), ("Bob", 88)]
        for student, score in tasks:
            p = Process(target=worker_scores, args=(shared_data, student, score))
            processes.append(p)
            p.start()
        for p in processes:
            p.join()
        print("Final shared student scores:", dict(shared_data))
```
