### Practice 3: Thread-Safe Counter
Use a lock to safely increment a shared counter.

```python

counter = 0
lock = threading.Lock()

def increment():
    global counter
    for _ in range(1000):
        with lock:
            counter += 1

# TODO: Start 5 threads that call increment. Wait for them to finish, then print the counter.
```

### Practice 3: Using a Queue
Pass numbers into a process and get back their squares.

```python

def square_numbers(q_in, q_out):
    while not q_in.empty():
        num = q_in.get()
        q_out.put(num * num)

# TODO: Put numbers 1–5 in q_in, start a process to run square_numbers, collect results.
```
